Defining multidimensional arrays using TensorFlow

```python
scalar = tf.constant(2)
vector = tf.constant([5,6,2])
matrix = tf.constant([[1,2,3],[2,3,4],[3,4,5]])
tensor = tf.constant([ [[1,2,3],[2,3,4],[3,4,5]], [[4,5,6],[5,6,7],[6,7,8]], [[7,8,9],[8,9,10],[9,10,11]] ])

print ("Scalar (1 entry):\n %s \n" % scalar)

print ("Vector (3 entries) :\n %s \n" % vector)

print ("Matrix (3x3 entries):\n %s \n" % matrix)

print ("Tensor (3x3x3 entries) :\n %s \n" % tensor)
```

tf.shape returns the shape of our data structure

```python
print('shape of scalar is %s\n', scalar.shape)
print('shape of scalar is %s\n', vector.shape)
print('shape of scalar is %s\n', matrix.shape)
print('shape of scalar is %s\n', tensor.shape)
```

```python
matrix_one = tf.constant([[1,2,3],[2,3,4],[3,4,5]])
matrix_two = tf.constant([[2,2,2],[2,2,2],[2,2,2]])

@tf.function
def add(a, b):
    res = tf.add(a, b)
    return res


print ("Defined using tensorflow function :")
add_1_operation = add(matrix_one, matrix_two)
print(add_1_operation)
print ("Defined using normal expressions :")
add_2_operation = matrix_one + matrix_two
print(add_2_operation)
```

```python
@tf.function
def matmul(a, b):
    res = tf.matmul(a, b)
    return res


mul_operation = matmul(matrix_one, matrix_two)

print ("Defined using tensorflow function :")
print(mul_operation)
```

```python
@tf.function
def increment_by_one(v):
    v = tf.add(v,1)
    return v
```

```python
for i in range(3):
    w = increment_by_one(w)
    print(w.numpy())
```
